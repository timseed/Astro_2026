```python
import pandas as pa
from pyongc import ongc
from pyongc import data

# This initial CSV File have the NGC id, and some other data
# I will use this and merge with a bigger and more complete data filtering as we go
df=pa.read_csv("herschel_400.csv",sep=",")
df['CAT_ID']=df['CAT_ID'].astype(int)
df['name']=df['CAT_ID'].apply(lambda x: f"NGC{x:04d}")
#We now have a nicely formatted NGCnnnn String
#Name is a common field used later

large_df=data.all()
ngc_df=large_df[large_df.name.str.contains("NGC")]
# Lets inner join (i.e. make a smaller result) using the name field in both data frames
h_400_df = pa.merge(ngc_df, df, on='name', how='inner')

#Finally save the more complete but h400 data only
h_400_df.to_csv("Hers_400_full.csv")
```
