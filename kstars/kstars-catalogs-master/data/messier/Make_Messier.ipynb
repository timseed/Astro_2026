```python
from pyongc import data
import math
```

```python
df=data.all()
```

```python
KSTARS_TYPE_MAP = {
    # Stars
    "STAR": 1,
    "S": 1,
    "TYCHO": 1,
    # Solar system
    "PLANET": 2,
    "ASTEROID": 10,
    "COMET": 9,
    "SATELLITE": 16,
    # Clusters
    "OPEN": 3,
    "OPEN_CLUSTER": 3,
    "OC": 3,
    "GLOBULAR": 4,
    "GC": 4,
    "GLOBULAR_CLUSTER": 4,
    # Nebulae
    "GASEOUS": 5,
    "HII": 5,
    "DIFFUSE": 5,
    "EMISSION": 5,
    "PLANETARY": 6,
    "PN": 6,
    "SUPERNOVA_REMNANT": 7,
    "SNR": 7,
    # Galaxies
    "GALAXY": 8,
    "G": 8,
    "CLUSTER": 14,  # galaxy clusters
    "QUASAR": 15,
}

pyongc_mapping_to_kstars = {
         "G":KSTARS_TYPE_MAP["GALAXY"],
        "OCl":KSTARS_TYPE_MAP["OPEN_CLUSTER"],
        "Dup":KSTARS_TYPE_MAP["GALAXY"],  #Spiral 
        "*":KSTARS_TYPE_MAP["STAR"],
        "Other":KSTARS_TYPE_MAP["STAR"], # Can be Star, Nothing here 
        "**":KSTARS_TYPE_MAP["STAR"], # Double as in ** funny eh ? 
        "GPair":KSTARS_TYPE_MAP["CLUSTER"], #Galatic Pair ... so cluster ??
        "GCl":KSTARS_TYPE_MAP["CLUSTER"],
        "PN":KSTARS_TYPE_MAP["PLANETARY"],
        "Neb":KSTARS_TYPE_MAP["GASEOUS"],
        "HII":KSTARS_TYPE_MAP["HII"],
        "Cl+N":KSTARS_TYPE_MAP["DIFFUSE"],  #Star Cluster and Nebulae 
        "*Ass":KSTARS_TYPE_MAP["STAR"], #Loose cluster or stars
        "RfN":KSTARS_TYPE_MAP["GASEOUS"], #Reflection Nebulae
        "GTrpl":KSTARS_TYPE_MAP["CLUSTER"], #Galactic Triplet
        "GGroup":KSTARS_TYPE_MAP["CLUSTER"],
        "SNR":KSTARS_TYPE_MAP["SNR"],
        "NonEx":KSTARS_TYPE_MAP["STAR"], #Non Existent ... 
        "EmN":KSTARS_TYPE_MAP["EMISSION"],
        "Nova": KSTARS_TYPE_MAP["STAR"],#Nova Star
        "DrkN": KSTARS_TYPE_MAP["GASEOUS"], #Dark Nebulae
        }
df["KStars_Type"] = df["type"].map(pyongc_mapping_to_kstars)
df["ra_degrees"] = df.ra * 180.0 / math.pi
df["dec_degrees"] = df["dec"].apply(math.degrees)
```

```python
df_messier=df[df['messier'].str.len() > 0]
```

```python
df_messier.to_csv("Messier_from_pyongc.csv")
```
